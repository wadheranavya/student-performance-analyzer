```python
student_name = "Aarav"

maths = 85
science = 90
english = 80

total = maths + science + english
average = total / 3
percentage = (total / 300) * 100

if average >= 90:
    grade = "A+"
elif average >= 80:
    grade = "A"
elif average >= 70:
    grade = "B"
elif average >= 60:
    grade = "C"
else:
    grade = "Needs Improvement"

if maths >= science and maths >= english:
    best_subject = "Maths"
elif science >= maths and science >= english:
    best_subject = "Science"
else:
    best_subject = "English"

if average >= 90:
    remark = "Excellent performance!"
elif average >= 75:
    remark = "Good performance!"
elif average >= 60:
    remark = "Keep working hard!"
else:
    remark = "More practice needed."

print("STUDENT PERFORMANCE REPORT")
print("--------------------------")
print("Student Name:", student_name)
print("Total Marks:", total, "/ 300")
print("Average Marks:", average)
print("Percentage:", percentage, "%")
print("Grade:", grade)
print("Best Subject:", best_subject)
print("Remark:", remark)
```

```python
import pandas as pd

data = {
    "Student": [
        "Aarav", "Diya", "Kabir", "Meera", "Rohan",
        "Anaya", "Arjun", "Ishita", "Vivaan", "Sara"
    ],
    "Maths": [85, 92, 95, 78, 70, 88, 81, 96, 74, 90],
    "Science": [90, 89, 94, 85, 72, 91, 84, 93, 80, 87],
    "English": [80, 95, 88, 90, 75, 86, 89, 91, 82, 94]
}

df = pd.DataFrame(data)

df["Total"] = df[["Maths", "Science", "English"]].sum(axis=1)
df["Average"] = df[["Maths", "Science", "English"]].mean(axis=1)
df["Percentage"] = df["Average"]

print("STUDENT PERFORMANCE DATASET")
display(df)
```

```python
# Step 2: Analyze student performance

# Assign grades based on average marks
def assign_grade(average):
    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    else:
        return "Needs Improvement"


df["Grade"] = df["Average"].apply(assign_grade)

# Rank students by average marks
df["Rank"] = df["Average"].rank(
    ascending=False,
    method="min"
).astype(int)

# Find the highest and lowest average scores
top_student = df.loc[df["Average"].idxmax()]
lowest_student = df.loc[df["Average"].idxmin()]

# Calculate class statistics
class_average = df["Average"].mean()
highest_average = df["Average"].max()
lowest_average = df["Average"].min()

print("STUDENT PERFORMANCE ANALYSIS")
print("-----------------------------")
print(f"Class average: {class_average:.2f}%")
print(f"Highest average: {highest_average:.2f}%")
print(f"Lowest average: {lowest_average:.2f}%")
print(f"Top student: {top_student['Student']}")
print(f"Lowest-scoring student: {lowest_student['Student']}")

print("\nSTUDENT RANKINGS")
display(
    df.sort_values("Rank")[
        ["Rank", "Student", "Average", "Grade"]
    ]
)
```

```python
# STUDENT PERFORMANCE ANALYZER
# A beginner Data Science portfolio project

import pandas as pd
import matplotlib.pyplot as plt

# 1. Create a sample dataset
data = {
    "Student": [
        "Aarav", "Diya", "Kabir", "Meera", "Rohan",
        "Anaya", "Arjun", "Ishita", "Vivaan", "Sara"
    ],
    "Maths": [85, 92, 95, 78, 70, 88, 81, 96, 74, 90],
    "Science": [90, 89, 94, 85, 72, 91, 84, 93, 80, 87],
    "English": [80, 95, 88, 90, 75, 86, 89, 91, 82, 94]
}

df = pd.DataFrame(data)

# 2. Calculate total, average, and percentage
subjects = ["Maths", "Science", "English"]

df["Total"] = df[subjects].sum(axis=1)
df["Average"] = df[subjects].mean(axis=1)
df["Percentage"] = df["Average"]

# 3. Assign grades
def assign_grade(average):
    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    else:
        return "Needs Improvement"

df["Grade"] = df["Average"].apply(assign_grade)

# 4. Rank students
df["Rank"] = (
    df["Average"]
    .rank(ascending=False, method="min")
    .astype(int)
)

# 5. Display the full dataset
print("STUDENT PERFORMANCE REPORT")
print("=" * 35)
display(df.sort_values("Rank"))

# 6. Display class-level insights
print("\nCLASS INSIGHTS")
print("=" * 35)
print(f"Number of students: {len(df)}")
print(f"Class average: {df['Average'].mean():.2f}%")
print(f"Highest average: {df['Average'].max():.2f}%")
print(f"Lowest average: {df['Average'].min():.2f}%")

top_student = df.loc[df["Average"].idxmax()]
lowest_student = df.loc[df["Average"].idxmin()]

print(f"Top performer: {top_student['Student']}")
print(f"Lowest average: {lowest_student['Student']}")

subject_averages = df[subjects].mean()
print("\nAverage by subject:")
display(subject_averages.rename("Average Marks").to_frame())

# 7. Chart: Average marks by subject
plt.figure(figsize=(8, 5))
subject_averages.plot(kind="bar")
plt.title("Average Marks by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Marks out of 100")
plt.ylim(0, 100)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# 8. Chart: Student performance comparison
ranked_students = df.sort_values("Average", ascending=False)

plt.figure(figsize=(10, 5))
plt.bar(ranked_students["Student"], ranked_students["Average"])
plt.title("Student Performance Comparison")
plt.xlabel("Student")
plt.ylabel("Average Marks (%)")
plt.ylim(0, 100)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 9. Chart: Subject scores for each student
df.set_index("Student")[subjects].plot(
    kind="bar",
    figsize=(12, 6)
)
plt.title("Subject-wise Marks for Each Student")
plt.xlabel("Student")
plt.ylabel("Marks out of 100")
plt.ylim(0, 100)
plt.xticks(rotation=45)
plt.legend(title="Subject")
plt.tight_layout()
plt.show()

# 10. Automatically generated findings
best_subject = subject_averages.idxmax()
needs_attention = subject_averages.idxmin()

print("\nKEY FINDINGS")
print("=" * 35)
print(f"Strongest subject overall: {best_subject}")
print(f"Lowest average subject: {needs_attention}")

if subject_averages.max() - subject_averages.min() < 5:
    print("Subject averages are relatively close.")
else:
    print("There is a noticeable difference between subject averages.")

print("\nProject completed successfully!")
```

# Student Performance Analyzer 📊

### A Beginner Data Science Portfolio Project

## Overview
The Student Performance Analyzer is a Python-based data analysis
project that explores academic marks across multiple subjects.

It uses Pandas to organize and analyze student data and Matplotlib
to create visualizations that make performance patterns easier
to understand.

## Project Objectives
- Calculate total marks, averages, percentages, and grades.
- Rank students based on their average marks.
- Compare Maths, Science, and English performance.
- Visualize subject averages and student results.
- Generate data-driven findings and recommendations.

## Tools Used
- Python
- Pandas
- Matplotlib
- Kaggle Notebooks

## Dataset
This project uses synthetic student data created for educational
purposes. The names and marks are illustrative.

## Learning Outcomes
This project demonstrates basic data cleaning and organization,
calculation, ranking, exploratory data analysis, visualization,
and interpretation of results.

## Limitations
The dataset contains only 10 illustrative students and three
subjects. The findings should not be generalized to real schools
or used alone to make decisions about individual students.
